```python

import os

from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings

api_key = os.environ.get("api_key", "xxx")

embedding_model = NVIDIAEmbeddings(
    model="nvidia/nemotron-3.5-embed-1b",
    nvidia_api_key=api_key
)
```

```python
from langchain_community.vectorstores.chroma import Chroma

vector = Chroma(persist_directory="D:\\rag_lectu\\.venv\\VectorDB", 
                embedding_function=embedding_model)
```

```python
context=vector.similarity_search(("display information about the data lakehouse ?"),k=3)
```

```python
from langchain_nvidia_ai_endpoints import ChatNVIDIA


llm = ChatNVIDIA(
  model="openai/gpt-oss-20b",
  api_key="xxx", 
  temperature=0,
  top_p=1,
  max_completion_tokens=4096,
)
```

```python
context=llm.invoke((f"display information about the data lakehouse with context: {context}"))
print(context.content)
```
