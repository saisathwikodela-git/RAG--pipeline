```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('D:\\rag_lectu\\.venv\\Document\\fabric-admin.pdf')
pdf = loader.load()

pdf
```

```python
for i in pdf:
    i.metadata={"source": "Dataengineer.pdf", "author": "Verma, Sumit"}
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter( chunk_size=1000, chunk_overlap=50)

chunks=splitter.split_documents(pdf)
chunks
```

## ** saving as chroms DB persist**

```python
import os
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings


api_key = os.environ.get("api_key", "xxx")

embedding_model = NVIDIAEmbeddings(
    model="nvidia/nemotron-3.5-embed-1b",
    nvidia_api_key=api_key
)
```

```python
from langchain_community.vectorstores.chroma import Chroma

vector= Chroma(
    persist_directory="D:\\rag_lectu\\.venv\\VectorDB",
    embedding_function=embedding_model)
```

```python
vector.add_documents(chunks)
```

```python
len(vector)
```

```python
vector.similarity_search(("what is microsoft admin ?"),k=3)
```
